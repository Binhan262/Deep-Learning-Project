## 1.1. Load and check dataset

```python
from gec_datasets import GECDatasets

# Initialize dataset manager
gec = GECDatasets(base_path="./data")

# Training datasets
wi_train = gec.load("wi-locness-train")
fce_train = gec.load("fce-train")

# Validation and testing
valid_data = gec.load("bea19-dev")
test_data = gec.load("bea19-test")

print("All datasets loaded successfully!")
```

```python
print("W&I+LOCNESS:", len(wi_train.srcs))
print("FCE:", len(fce_train.srcs))
print("BEA-2019 Dev:", len(valid_data.srcs))
print("BEA-2019 Test:", len(test_data.srcs))
```

```python
# W&I+LOCNESS example
print("Source:", wi_train.srcs[0])
print("Target:", wi_train.refs[0][0])

# FCE example
print("\nSource:", fce_train.srcs[0])
print("Target:", fce_train.refs[0][0])

# Validation example
print("\nValidation:", valid_data.srcs[0])
print("Reference:", valid_data.refs[0][0])

# Test example
print("\nTest:", test_data.srcs[0])
```

```python
print("W&I references:", len(wi_train.refs))
print("FCE references:", len(fce_train.refs))
print("Dev references:", len(valid_data.refs))
print("Test references:", len(test_data.refs))
```
